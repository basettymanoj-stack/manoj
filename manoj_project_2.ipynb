```python
# ================================================================
# 🚀 AI STUDY HUB - FINAL GOOGLE COLAB VERSION
# Login + Signup + Dashboard + 11 Topics + Study Material
# 10 MCQs/topic + Timer + Score + Prediction + SQLite
# Automatic Colab Website Link
# QUIZ INTERNAL SERVER ERROR FIXED
# ================================================================

!pip -q install flask flask-cors scikit-learn pandas numpy werkzeug

import os
import sqlite3
import threading
import time
import random
import socket
import json

import numpy as np
import pandas as pd

from flask import Flask, request, redirect, session, jsonify, render_template_string
from werkzeug.security import generate_password_hash, check_password_hash
from sklearn.ensemble import RandomForestRegressor

# ================================================================
# 1. DATABASE
# ================================================================

BASE_DIR = "/content/AI_Study_Hub_Final"
os.makedirs(BASE_DIR, exist_ok=True)

DB_PATH = os.path.join(BASE_DIR, "study_hub.db")

conn = sqlite3.connect(DB_PATH)
cur = conn.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS users(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    username TEXT UNIQUE NOT NULL,
    password TEXT NOT NULL
)
""")

cur.execute("""
CREATE TABLE IF NOT EXISTS results(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    user_id INTEGER,
    topic TEXT,
    score INTEGER,
    total INTEGER,
    percentage REAL,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
""")

conn.commit()
conn.close()

# ================================================================
# 2. STUDY MATERIAL + QUIZ DATA
# ================================================================

def make_questions(items):
    """
    items = [
      (question, [A,B,C,D], correct_index)
    ]
    """
    return items


topics = {

"Artificial Intelligence": {
"icon":"🤖",
"material":"""
Artificial Intelligence (AI) is a branch of computer science that enables machines and computer systems to perform tasks that normally require human intelligence. AI systems can learn from data, recognize patterns, understand language, solve problems, and make decisions.

AI works using techniques such as Machine Learning, Deep Learning, Natural Language Processing, Computer Vision, and Robotics. It is widely used in education, healthcare, banking, transportation, entertainment, and many other fields.

The main goal of AI is to create systems that can learn, reason, understand, and make decisions with minimum human intervention.

Example

Example: Voice Assistant

Google Assistant, Siri, and Alexa use AI to understand human voice commands and provide suitable responses.

Other examples:

Face recognition
Chatbots
Recommendation systems
Self-driving vehicles
Medical diagnosis systems
Advantages of AI
Reduces human effort – AI can perform repetitive tasks automatically.
Fast decision making – AI can process large amounts of data quickly.
24/7 availability – AI systems can work continuously without fatigue.
Improves accuracy – AI can reduce errors in many data-based tasks.
Automation – AI can automate complex and repetitive processes.
Disadvantages of AI
High cost – Developing and maintaining AI systems can be expensive.
Job displacement – Automation may reduce the need for some types of human work.
Privacy concerns – AI systems often process large amounts of personal data.
Lack of human emotions – AI cannot truly understand human feelings like people do.
Dependency on data – Poor-quality or biased data can produce poor results.
Simple Real-Life Example

Online Shopping Recommendation

When you search for a product on an online shopping website, AI analyzes your previous searches and purchases and recommends similar products.

Input → User Data → AI Model → Analysis → Recommended Products
""",
"questions": make_questions([

    (
        "What does AI stand for?",
        [
            "Automated Internet",
            "Artificial Intelligence",
            "Advanced Information",
            "Automatic Intelligence"
        ],
        1
    ),

    (
        "What is the main goal of AI?",
        [
            "Store files",
            "Create intelligent systems",
            "Increase internet speed",
            "Design websites"
        ],
        1
    ),

    (
        "Which of the following is an example of AI?",
        [
            "Calculator",
            "Voice Assistant",
            "Keyboard",
            "USB Drive"
        ],
        1
    ),

    (
        "Which AI technique allows computers to learn from data?",
        [
            "Machine Learning",
            "Networking",
            "Operating System",
            "Database"
        ],
        0
    ),

    (
        "Which of these is an AI voice assistant?",
        [
            "Alexa",
            "Windows Explorer",
            "Notepad",
            "Paint"
        ],
        0
    ),

    (
        "Which AI technology is used to understand human language?",
        [
            "NLP",
            "RAM",
            "HTML",
            "CPU"
        ],
        0
    ),

    (
        "Which is an advantage of AI?",
        [
            "Always needs human control",
            "Can automate repetitive tasks",
            "Cannot process data",
            "Works only during daytime"
        ],
        1
    ),

    (
        "Which is a disadvantage of AI?",
        [
            "Fast processing",
            "Automation",
            "High development cost",
            "24/7 operation"
        ],
        2
    ),

    (
        "Face recognition is an application of which technology?",
        [
            "Computer Vision",
            "Database Management",
            "Cloud Storage",
            "Web Design"
        ],
        0
    ),

    (
        "Which field commonly uses AI for disease detection?",
        [
            "Healthcare",
            "Sports only",
            "Music only",
            "Printing"
        ],
        0
    )

])
},
"Machine Learning": {
"icon":"🧠",
"material":"""
1. What is Machine Learning?

Machine Learning (ML) is a branch of Artificial Intelligence that enables computers to learn from data and make predictions or decisions without being explicitly programmed.

2. Types of Machine Learning

1. Supervised Learning

Learns from labelled data.
Used for prediction and classification.
Examples: Linear Regression, Decision Tree, SVM, KNN.

2. Unsupervised Learning

Learns from unlabelled data.
Finds patterns or groups in data.
Examples: K-Means, Hierarchical Clustering.

3. Reinforcement Learning

Learns by rewards and penalties.
Used for decision-making.
Example: Game-playing AI, robots.
3. Important ML Concepts
Concept	Simple Meaning
Dataset	Collection of data
Feature	Input variable/characteristic
Label	Expected output
Training Data	Data used to train the model
Testing Data	Data used to test the model
Model	Learned pattern used for prediction
Classification	Predicts categories
Regression	Predicts numerical values
Clustering	Groups similar data
Overfitting	Model learns training data too closely
Underfitting	Model fails to learn enough
Accuracy	Percentage of correct predictions
Loss	Measures prediction error
4. Common Machine Learning Algorithms
Linear Regression – predicts continuous values.
Logistic Regression – classification.
Decision Tree – uses tree-like decisions.
Random Forest – combines multiple decision trees.
KNN – predicts based on nearby data points.
SVM – finds the best boundary between classes.
K-Means – groups similar data.
Naive Bayes – probability-based classification.
Neural Networks – learns complex patterns.
5. Basic ML Workflow

Data Collection → Data Preprocessing → Feature Selection → Train Model → Test Model → Evaluate Model → Prediction

6. Example

Email Spam Detection

Input: Email text
Features: Words, frequency, etc.
Model: Naive Bayes
Output: Spam / Not Spam
""",
"questions":make_questions([
("What is the main goal of Machine Learning?",
 ["Learn from data", "Create websites", "Design hardware", "Send emails"], 0),

("Which type of learning uses labelled data?",
 ["Supervised Learning", "Unsupervised Learning", "Reinforcement Learning", "Random Learning"], 0),

("Which type of learning finds hidden patterns in data?",
 ["Unsupervised Learning", "Supervised Learning", "Reinforcement Learning", "Manual Learning"], 0),

("Which algorithm is mainly used for classification?",
 ["KNN", "K-Means", "Linear Regression", "PCA"], 0),

("Which algorithm is used for clustering?",
 ["K-Means", "Decision Tree", "Linear Regression", "Logistic Regression"], 0),

("What is a feature in Machine Learning?",
 ["An input variable", "The final answer", "A computer program", "A database"], 0),

("What is overfitting?",
 ["Model learns training data too closely",
  "Model does not use data",
  "Model deletes data",
  "Model creates new hardware"], 0),

("Which metric measures the percentage of correct predictions?",
 ["Accuracy", "Loss", "Speed", "Memory"], 0),

("What is used to evaluate a trained Machine Learning model?",
 ["Testing data", "Keyboard", "Monitor", "Operating System"], 0),

("What does reinforcement learning use to learn?",
 ["Rewards and penalties", "Only labelled data",
  "SQL commands", "Web pages"], 0)
])
},

"Data Science": {
"icon":"📊",
"material":"""
1. What is Data Science?

Data Science is the process of collecting, cleaning, analyzing, and interpreting data to find useful information and make decisions.

2. Main Steps in Data Science

Data Collection → Data Cleaning → Data Analysis → Data Visualization → Model Building → Evaluation → Decision Making

3. Important Concepts
Concept	Simple Meaning
Data	Collection of information
Dataset	Collection of related data
Data Collection	Gathering data from different sources
Data Cleaning	Removing errors and unwanted data
Data Preprocessing	Preparing data for analysis
Feature	An important input variable
Data Analysis	Finding useful information from data
Data Visualization	Representing data using charts/graphs
Statistics	Studying data using mathematical methods
Machine Learning	Learning patterns from data
Prediction	Estimating future or unknown results
Model	A system trained to analyze or predict data
4. Common Data Science Tools
Python – programming
NumPy – numerical calculations
Pandas – data handling
Matplotlib – data visualization
Seaborn – statistical graphs
Scikit-learn – machine learning
Jupyter Notebook / Google Colab – coding and analysis
5. Types of Data

1. Structured Data

Organized in rows and columns.
Example: Excel table.

2. Unstructured Data

No fixed structure.
Example: Images, videos, emails.

3. Semi-structured Data

Has some organization.
Example: JSON and XML.
6. Simple Example

Student Performance Analysis

Collect student marks.
Clean missing or incorrect values.
Calculate average marks.
Create a graph.
Find students who need improvement.
Use the results to make decisions.
""",
"questions":make_questions([
("What is the main purpose of Data Science?",
 ["Extract useful information from data",
  "Create computer hardware",
  "Repair networks",
  "Design keyboards"], 0),

("Which language is widely used in Data Science?",
 ["Python", "HTML", "CSS", "SQL only"], 0),

("Which library is commonly used to work with tables and datasets?",
 ["Pandas", "Flask", "Django", "Tkinter"], 0),

("Which library is useful for numerical operations?",
 ["NumPy", "Pandas", "Flask", "BeautifulSoup"], 0),

("Which Python library is commonly used to create graphs?",
 ["Matplotlib", "NumPy", "Requests", "OS"], 0),

("What is data preprocessing?",
 ["Preparing data for analysis",
  "Installing an operating system",
  "Creating a website",
  "Connecting a printer"], 0),

("What does data cleaning mainly involve?",
 ["Removing or correcting incorrect data",
  "Increasing computer speed",
  "Installing software",
  "Creating passwords"], 0),

("What is Exploratory Data Analysis (EDA)?",
 ["Studying data to find patterns and relationships",
  "Building computer networks",
  "Writing operating systems",
  "Designing hardware"], 0),

("Which tool can be used to build Machine Learning models?",
 ["Scikit-learn", "HTML", "CSS", "FTP"], 0),

("Why is data visualization useful?",
 ["It makes data easier to understand",
  "It increases RAM",
  "It repairs computers",
  "It connects networks"], 0)

])
},

"Computer Networks": {
"icon":"🌐",
"material":"""
A computer network is a collection of interconnected devices that communicate
and share resources. Network communication uses protocols that define how
data is transmitted. The OSI model contains seven layers: Physical, Data
Link, Network, Transport, Session, Presentation and Application. TCP and UDP
are important transport-layer protocols. IP provides logical addressing and
routing. DNS translates domain names into IP addresses. Routers forward
packets between networks while switches connect devices within a local
network. Networks are essential for the Internet, cloud services and
distributed applications.
""",
"questions":make_questions([
("How many layers are in the OSI model?",["7","5","4","10"],0),
("Which protocol provides reliable transport?",["TCP","UDP","IP","DNS"],0),
("Which protocol translates domain names?",["DNS","FTP","ARP","ICMP"],0),
("A router connects?",["Different networks","Only keyboards","Only printers","Only monitors"],0),
("A switch commonly connects devices in?",["LAN","Only WAN","Only satellites","Only CPUs"],0),
("IP is used for?",["Logical addressing and routing","Printing","Text editing","Video rendering"],0),
("UDP is?",["Connectionless","Always connection-oriented","A database","An operating system"],0),
("Which OSI layer handles routing?",["Network","Physical","Presentation","Session"],0),
("HTTP is mainly used for?",["Web communication","Routing hardware","Disk formatting","CPU scheduling"],0),
("A packet contains?",["Network data and control information","Only electricity","Only a password","Only a screen"],0)
])
},

"Cyber Security": {
"icon":"🔐",
"material":"""
Cyber security protects computers, networks, applications and data from
unauthorized access, attacks and damage. Important security goals are
confidentiality, integrity and availability, commonly called the CIA triad.
Authentication verifies identity while authorization determines what an
authenticated user can access. Firewalls control network traffic and
antivirus software detects malicious programs. Encryption protects data by
converting plaintext into ciphertext. Strong passwords, multi-factor
authentication, regular updates and secure backups improve security.
""",
"questions":make_questions([
("What does CIA stand for in security?",["Confidentiality, Integrity, Availability","Computer Internet Access","Control Internet Application","Central Information Area"],0),
("Authentication verifies?",["Identity","Printer status","CPU speed","File size"],0),
("Authorization determines?",["Access permissions","Password length only","Screen size","Network speed"],0),
("A firewall controls?",["Network traffic","Keyboard lights","CPU temperature","Monitor resolution"],0),
("Encryption converts plaintext into?",["Ciphertext","HTML","IP address","Database"],0),
("Malware means?",["Malicious software","Mail software","Manual software","Machine language"],0),
("MFA improves?",["Authentication security","Screen brightness","CPU speed","Disk capacity"],0),
("A strong password should be?",["Difficult to guess","Very short","Public","Same everywhere"],0),
("A backup helps protect against?",["Data loss","Screen damage","Keyboard failure only","Low battery"],0),
("Phishing commonly attempts to steal?",["Sensitive information","Monitor size","CPU speed","Printer ink"],0)
])
},

"Python Programming": {
"icon":"🐍",
"material":"""
Python is a high-level programming language known for readable syntax and a
large ecosystem of libraries. It supports procedural, object-oriented and
functional programming styles. Variables can store values without explicit
type declarations. Common Python data structures include lists, tuples, sets
and dictionaries. Functions improve code organization and reuse. Libraries
such as NumPy, Pandas, Flask and Scikit-learn make Python useful for data
science, web development and machine learning.
""",
"questions":make_questions([
("Python is a?",["Programming language","Database","Operating system","Router"],0),
("Which symbol starts a comment in Python?",["#","//","<!--","**"],0),
("Which data type stores key-value pairs?",["Dictionary","List","Tuple","String"],0),
("Which keyword defines a function?",["def","function","fun","define"],0),
("Which collection is ordered and mutable?",["List","Tuple","Set","None"],0),
("Which library is used for numerical arrays?",["NumPy","Flask","Django","HTML"],0),
("Pandas is mainly used for?",["Data analysis","Routing","Encryption","Gaming"],0),
("Which keyword is used for a loop?",["for","loop","repeat","iterate"],0),
("Python files commonly use?",[".py",".html",".sql",".exeonly"],0),
("Flask is used for?",["Web development","Disk formatting","Routing hardware","Antivirus"],0)
])
},

"Database Management Systems": {
"icon":"🗄️",
"material":"""
A Database Management System (DBMS) is software used to store, organize,
retrieve and manage data. Relational databases store information in tables
containing rows and columns. SQL is used to create, modify and query
relational data. Primary keys uniquely identify records, while foreign keys
represent relationships between tables. Database normalization reduces
redundancy and improves consistency. Transactions follow properties such as
atomicity, consistency, isolation and durability, known as ACID.
""",
"questions":make_questions([
("DBMS is used to?",["Manage data","Edit images","Connect monitors","Compile hardware"],0),
("SQL stands for?",["Structured Query Language","Simple Question Language","System Query Link","Sequential Query Logic"],0),
("A table contains?",["Rows and columns","Only images","Only programs","Only passwords"],0),
("A primary key uniquely identifies?",["A record","A database server","A keyboard","A network"],0),
("A foreign key represents?",["A relationship","A password","A CPU","A file system"],0),
("Normalization reduces?",["Data redundancy","Internet speed","RAM","Screen size"],0),
("ACID relates to?",["Transactions","Networking cables","Graphics","Operating systems"],0),
("Which command retrieves data?",["SELECT","DELETE","DROP","UPDATE"],0),
("Which command adds records?",["INSERT","SELECT","CREATE","ALTER"],0),
("A relational database stores data in?",["Tables","Routers","Packets","Processes"],0)
])
},

"Operating Systems": {
"icon":"💻",
"material":"""
An operating system is system software that manages computer hardware and
provides services for application programs. Important operating-system
functions include process management, memory management, file management,
device management and security. A process is a program in execution. CPU
scheduling decides which ready process gets the processor. Virtual memory
allows systems to use secondary storage as an extension of main memory.
Operating systems also provide file systems and interfaces through which
users interact with computers.
""",
"questions":make_questions([
("An operating system manages?",["Hardware and software resources","Only websites","Only databases","Only emails"],0),
("A process is?",["A program in execution","A file only","A printer","A network"],0),
("CPU scheduling selects?",["A process to run","A password","A file format","A website"],0),
("Virtual memory uses?",["Secondary storage","Only cache","Only keyboard","Only monitor"],0),
("Which manages files?",["File system","Router","Compiler","Browser"],0),
("Which is a CPU scheduling algorithm?",["Round Robin","RSA","DNS","HTTP"],0),
("Deadlock involves?",["Processes waiting indefinitely","Fast execution","Printing","Web browsing"],0),
("Memory management handles?",["Allocation of memory","Email delivery","Web pages","Network routing"],0),
("A device driver helps OS communicate with?",["Hardware devices","Only users","Only databases","Only websites"],0),
("Linux is an example of?",["Operating system","Database","Browser","Protocol"],0)
])
},

"Cloud Computing": {
"icon":"☁️",
"material":"""
Cloud computing provides computing resources such as servers, storage,
databases, networking and software through the Internet. The major service
models are Infrastructure as a Service, Platform as a Service and Software
as a Service. Deployment models include public, private, hybrid and
community clouds. Virtualization allows multiple virtual machines to share
physical resources. Cloud computing offers scalability, flexibility and
pay-as-you-use resource models. It is widely used for web hosting,
application development, data storage and machine learning.
""",
"questions":make_questions([
("Cloud computing provides resources through?",["Internet","Keyboard","Printer","BIOS"],0),
("IaaS stands for?",["Infrastructure as a Service","Internet as a Service","Information as Software","Interface as a Service"],0),
("PaaS provides?",["Application development platform","Only hardware cables","Only email","Only antivirus"],0),
("SaaS provides?",["Software as a service","Only processors","Only networks","Only storage devices"],0),
("Which is a cloud deployment model?",["Public cloud","CPU cloud","Keyboard cloud","Printer cloud"],0),
("Hybrid cloud combines?",["Public and private clouds","Two keyboards","Two CPUs","Two browsers"],0),
("Virtualization creates?",["Virtual resources","Only documents","Only passwords","Only emails"],0),
("Cloud scalability means?",["Resources can increase or decrease","Screen becomes larger","Keyboard changes","Files disappear"],0),
("Cloud storage stores?",["Data remotely","Only CPU instructions","Only passwords","Only cables"],0),
("Docker is commonly associated with?",["Containers","Word processing","DNS only","Printers"],0)
])
},

"Computer Science Fundamentals": {
"icon":"🧩",
"material":"""
Computer science fundamentals include algorithms, data structures, computer
architecture, programming, databases, operating systems and networking.
Algorithms are step-by-step procedures for solving problems. Data structures
organize information so that operations can be performed efficiently.
Examples include arrays, linked lists, stacks, queues, trees and graphs.
Complexity analysis evaluates time and space requirements. Understanding
these fundamentals helps students design efficient software and solve
programming problems systematically.
""",
"questions":make_questions([
("An algorithm is?",["Step-by-step solution","A hardware device","A database","A network cable"],0),
("A stack follows?",["LIFO","FIFO","Random only","Circular only"],0),
("A queue follows?",["FIFO","LIFO","Random","Binary"],0),
("A tree is a?",["Data structure","Protocol","Database command","Programming language"],0),
("A graph contains?",["Vertices and edges","Rows only","Columns only","Passwords"],0),
("Complexity measures?",["Resource requirements","Screen size","Keyboard speed","Network color"],0),
("Binary search requires data to be?",["Sorted","Encrypted","Compressed","Printed"],0),
("An array stores?",["Elements in indexed positions","Only passwords","Only networks","Only files"],0),
("A linked list contains?",["Nodes and links","Only rows","Only packets","Only images"],0),
("Which structure is useful for BFS?",["Queue","Stack only","Tree only","Array only"],0)
])
},

"Web Development": {
"icon":"🌍",
"material":"""
Web development involves creating websites and web applications. Front-end
development focuses on what users see and interact with, using technologies
such as HTML, CSS and JavaScript. Back-end development handles server-side
logic, databases and APIs. Flask is a Python web framework that can be used
to build web applications. HTML provides structure, CSS controls presentation
and JavaScript adds interactivity. Modern web applications often use APIs to
exchange data between the browser and server.
""",
"questions":make_questions([
("HTML is used for?",["Web page structure","Database storage","CPU scheduling","Encryption"],0),
("CSS controls?",["Presentation and styling","Database queries","Network routing","Passwords"],0),
("JavaScript provides?",["Interactivity","Only storage","Only encryption","Only routing"],0),
("Flask is a?",["Python web framework","Database","Browser","Operating system"],0),
("Backend handles?",["Server-side logic","Only colors","Only fonts","Only images"],0),
("Frontend runs mainly in?",["Browser","Router","Hard disk","BIOS"],0),
("API is used for?",["Communication between software components","Screen brightness","Disk formatting","Keyboard input only"],0),
("HTTP is a?",["Web communication protocol","Programming language","Database","Hardware device"],0),
("JSON is commonly used for?",["Data exchange","CPU scheduling","Image editing","Disk partitioning"],0),
("A web server handles?",["Web requests","Only printing","Only keyboard input","Only battery charging"],0)
])
}

}

# ================================================================
# 3. ML SCORE PREDICTION MODEL
# ================================================================

X_train = np.array([
    [1,20,40],[2,25,45],[3,30,50],[4,35,55],[5,40,60],
    [6,45,65],[7,50,70],[8,55,75],[9,60,80],[10,65,90]
])

y_train = np.array([35,40,45,50,55,60,68,75,82,90])

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

# ================================================================
# 4. FLASK APP
# ================================================================

app = Flask(__name__)
app.secret_key = "AI_STUDY_HUB_SECRET_2026"

# ================================================================
# 5. COMMON CSS
# ================================================================

COMMON_CSS = """
<style>

*{
    box-sizing:border-box;
}

body{
    margin:0;
    font-family:Arial,Helvetica,sans-serif;
    min-height:100vh;
    color:white;
    background:linear-gradient(135deg,#4f46e5,#06b6d4,#9333ea);
    background-size:400% 400%;
    animation:bgMove 12s ease infinite;
}

@keyframes bgMove{
    0%{background-position:0% 50%}
    50%{background-position:100% 50%}
    100%{background-position:0% 50%}
}

.container{
    width:92%;
    max-width:1200px;
    margin:auto;
}

.card{
    background:rgba(255,255,255,.13);
    backdrop-filter:blur(18px);
    border:1px solid rgba(255,255,255,.2);
    border-radius:22px;
    padding:25px;
    box-shadow:0 15px 40px rgba(0,0,0,.2);
}

input{
    width:100%;
    padding:14px;
    margin:8px 0;
    border:none;
    border-radius:12px;
    outline:none;
    font-size:16px;
}

button,.btn{
    border:none;
    border-radius:12px;
    padding:13px 20px;
    cursor:pointer;
    font-weight:bold;
    text-decoration:none;
    display:inline-block;
    background:white;
    color:#4f46e5;
}

button:hover,.btn:hover{
    transform:translateY(-2px);
}

.nav{
    display:flex;
    justify-content:space-between;
    align-items:center;
    padding:20px 0;
    flex-wrap:wrap;
    gap:10px;
}

.grid{
    display:grid;
    grid-template-columns:repeat(auto-fit,minmax(260px,1fr));
    gap:20px;
}

.topic-card{
    transition:.3s;
}

.topic-card:hover{
    transform:translateY(-8px);
}

.icon{
    font-size:45px;
}

h1{
    font-size:40px;
}

h2{
    font-size:28px;
}

h3{
    font-size:21px;
}

p{
    line-height:1.8;
}

.badge{
    display:inline-block;
    padding:7px 12px;
    background:rgba(255,255,255,.2);
    border-radius:20px;
    margin:4px;
}

.stat{
    text-align:center;
    font-size:30px;
    font-weight:bold;
}

.center{
    min-height:100vh;
    display:flex;
    justify-content:center;
    align-items:center;
}

.auth{
    width:95%;
    max-width:430px;
    text-align:center;
}

.option{
    display:block;
    padding:15px;
    margin:10px 0;
    background:rgba(255,255,255,.12);
    border:1px solid rgba(255,255,255,.2);
    border-radius:12px;
    cursor:pointer;
}

.option:hover{
    background:rgba(255,255,255,.25);
}

.option input{
    width:auto;
    margin-right:10px;
}

.timer{
    position:sticky;
    top:10px;
    z-index:10;
    text-align:center;
    font-size:25px;
    font-weight:bold;
    background:#ef4444;
    padding:12px;
    border-radius:15px;
    margin-bottom:20px;
}

.material{
    font-size:17px;
    white-space:pre-line;
}

.result{
    text-align:center;
    padding:30px;
}

.error{
    background:#7f1d1d;
    padding:20px;
    border-radius:15px;
    margin:20px 0;
}

@media(max-width:600px){
    h1{font-size:30px;}
    h2{font-size:24px;}
}

</style>
"""

# ================================================================
# 6. LOGIN
# ================================================================

LOGIN_HTML = COMMON_CSS + """

<style>

/* ================================================================
   LOGIN BACKGROUND
   ================================================================ */

body {
    overflow-x: hidden;
    min-height: 100vh;
    background: linear-gradient(135deg,#4f46e5,#06b6d4);
    transition: background 1s ease;
}

/* ================================================================
   ANIMATED BUBBLES
   ================================================================ */

.login-bubbles {
    position: fixed;
    inset: 0;
    overflow: hidden;
    pointer-events: none;
    z-index: 0;
}

.login-bubbles span {
    position: absolute;
    display: block;
    border-radius: 50%;
    background: rgba(255,255,255,0.12);
    border: 1px solid rgba(255,255,255,0.18);
    animation: bubbleFloat 12s infinite ease-in-out;
}

/* Different bubbles */

.login-bubbles span:nth-child(1) {
    width: 180px;
    height: 180px;
    left: 5%;
    top: 10%;
    animation-duration: 14s;
}

.login-bubbles span:nth-child(2) {
    width: 90px;
    height: 90px;
    left: 25%;
    top: 65%;
    animation-duration: 10s;
    animation-delay: 1s;
}

.login-bubbles span:nth-child(3) {
    width: 240px;
    height: 240px;
    right: 3%;
    top: -50px;
    animation-duration: 16s;
}

.login-bubbles span:nth-child(4) {
    width: 120px;
    height: 120px;
    right: 25%;
    bottom: 5%;
    animation-duration: 11s;
    animation-delay: 2s;
}

.login-bubbles span:nth-child(5) {
    width: 55px;
    height: 55px;
    left: 42%;
    top: 15%;
    animation-duration: 8s;
}

.login-bubbles span:nth-child(6) {
    width: 150px;
    height: 150px;
    left: 45%;
    bottom: -60px;
    animation-duration: 13s;
}

@keyframes bubbleFloat {

    0% {
        transform: translateY(0px) translateX(0px) scale(1);
    }

    50% {
        transform: translateY(-60px) translateX(30px) scale(1.08);
    }

    100% {
        transform: translateY(0px) translateX(0px) scale(1);
    }

}


/* ================================================================
   MAIN LOGIN CENTER
   ================================================================ */

.center {
    min-height: 100vh;
    width: 100%;
    display: flex;
    align-items: center;
    justify-content: center;

    padding: 30px;
    box-sizing: border-box;

    position: relative;
    z-index: 2;
}


/* ================================================================
   LANDSCAPE LOGIN CARD
   ================================================================ */

.card.auth {
    width: 1100px;
    max-width: 95vw;
    min-height: 570px;

    padding: 0;

    display: grid;
    grid-template-columns: 42% 58%;

    overflow: hidden;

    border-radius: 30px;

    background: rgba(255,255,255,0.12);

    border: 1px solid rgba(255,255,255,0.25);

    box-shadow:
        0 25px 70px rgba(0,0,0,0.25);

    backdrop-filter: blur(18px);
    -webkit-backdrop-filter: blur(18px);
}


/* ================================================================
   LEFT GRADUATION CAP SECTION
   ================================================================ */

.login-left {
    position: relative;

    display: flex;
    align-items: center;
    justify-content: center;

    min-height: 570px;

    background:
        radial-gradient(
            circle at 50% 50%,
            rgba(255,255,255,0.18),
            transparent 55%
        );

    overflow: hidden;
}


/* Big glowing circle behind cap */

.login-left::before {
    content: "";

    position: absolute;

    width: 330px;
    height: 330px;

    border-radius: 50%;

    background:
        radial-gradient(
            circle,
            rgba(168,85,247,0.75),
            rgba(59,130,246,0.18),
            transparent 70%
        );

    animation: capGlow 4s infinite ease-in-out;
}


@keyframes capGlow {

    0%,100% {
        transform: scale(1);
        opacity: 0.75;
    }

    50% {
        transform: scale(1.12);
        opacity: 1;
    }

}


/* Graduation cap */

.graduation-cap {
    position: relative;
    z-index: 3;

    font-size: 150px;

    filter:
        drop-shadow(0 20px 25px rgba(0,0,0,0.35));

    animation: capFloat 4s infinite ease-in-out;
}


@keyframes capFloat {

    0%,100% {
        transform: translateY(0px) rotate(-2deg);
    }

    50% {
        transform: translateY(-18px) rotate(2deg);
    }

}


/* Small decorative dots */

.cap-dot {
    position: absolute;

    width: 10px;
    height: 10px;

    border-radius: 50%;

    background: rgba(255,255,255,0.7);

    animation: dotPulse 3s infinite ease-in-out;
}

.dot1 {
    left: 20%;
    top: 25%;
}

.dot2 {
    right: 18%;
    top: 35%;
    animation-delay: 1s;
}

.dot3 {
    left: 25%;
    bottom: 25%;
    animation-delay: 2s;
}

@keyframes dotPulse {

    0%,100% {
        transform: scale(1);
        opacity: 0.4;
    }

    50% {
        transform: scale(1.7);
        opacity: 1;
    }

}


/* ================================================================
   RIGHT LOGIN SECTION
   ================================================================ */

.login-right {
    display: flex;
    flex-direction: column;

    justify-content: center;

    padding: 55px 65px;

    box-sizing: border-box;

    background: rgba(255,255,255,0.08);
}


/* Title */

.login-title {
    font-size: 52px;
    font-weight: 800;

    color: white;

    margin: 0 0 12px 0;

    text-align: center;

    letter-spacing: -1px;
}


/* Subtitle */

.login-subtitle {
    color: white;

    font-size: 20px;

    text-align: center;

    margin-bottom: 38px;

    opacity: 0.95;
}


/* ================================================================
   INPUT CONTAINER
   ================================================================ */

.login-input-box {
    position: relative;

    width: 100%;

    margin-bottom: 20px;
}


/* Input */

.login-input-box input {
    width: 100%;

    height: 72px;

    box-sizing: border-box;

    border: none;

    outline: none;

    border-radius: 18px;

    padding: 0 55px;

    font-size: 19px;

    background: rgba(255,255,255,0.85);

    color: #222;

    box-shadow:
        0 8px 25px rgba(0,0,0,0.08);

    transition: 0.3s;
}


.login-input-box input:focus {
    background: white;

    transform: translateY(-2px);

    box-shadow:
        0 10px 30px rgba(255,255,255,0.25);
}


/* Left input icon */

.input-icon {
    position: absolute;

    left: 19px;

    top: 50%;

    transform: translateY(-50%);

    font-size: 22px;

    z-index: 5;
}


/* ================================================================
   PASSWORD EYE
   ================================================================ */

.password-eye {
    position: absolute;

    right: 18px;

    top: 50%;

    transform: translateY(-50%);

    font-size: 22px;

    cursor: pointer;

    z-index: 10;

    user-select: none;

    transition: 0.2s;
}

.password-eye:hover {
    transform:
        translateY(-50%)
        scale(1.15);
}


/* ================================================================
   LOGIN BUTTON
   ================================================================ */

.login-button {
    width: 180px;

    height: 65px;

    border: none;

    border-radius: 18px;

    background: white;

    color: #4f46e5;

    font-size: 19px;

    font-weight: 800;

    cursor: pointer;

    display: block;

    margin: 12px auto 0;

    box-shadow:
        0 10px 25px rgba(0,0,0,0.15);

    transition: 0.3s;
}


.login-button:hover {
    transform: translateY(-4px) scale(1.03);

    box-shadow:
        0 15px 35px rgba(0,0,0,0.25);
}


/* Create account */

.create-account {
    text-align: center;

    color: white;

    font-size: 17px;

    margin-top: 30px;
}


.create-account a {
    color: white !important;

    font-weight: 800;

    text-decoration: underline;
}


/* Error */

.error {
    text-align: center;

    margin-top: 15px;
}


/* ================================================================
   MOBILE
   ================================================================ */

@media (max-width: 800px) {

    .card.auth {
        width: 95vw;

        min-height: auto;

        display: block;
    }

    .login-left {
        min-height: 220px;
    }

    .graduation-cap {
        font-size: 100px;
    }

    .login-right {
        padding: 35px 25px;
    }

    .login-title {
        font-size: 38px;
    }

    .login-subtitle {
        font-size: 16px;
    }

}


/* ================================================================
   SMALL MOBILE
   ================================================================ */

@media (max-width: 500px) {

    .center {
        padding: 10px;
    }

    .login-left {
        min-height: 180px;
    }

    .graduation-cap {
        font-size: 80px;
    }

    .login-right {
        padding: 25px 18px;
    }

    .login-title {
        font-size: 32px;
    }

    .login-input-box input {
        height: 60px;
        font-size: 16px;
    }

}

</style>


<!-- ============================================================
     ANIMATED BACKGROUND BUBBLES
     ============================================================ -->

<div class="login-bubbles">

    <span></span>
    <span></span>
    <span></span>
    <span></span>
    <span></span>
    <span></span>

</div>


<!-- ============================================================
     LOGIN MAIN
     ============================================================ -->

<div class="center">

<div class="card auth">


<!-- ============================================================
     LEFT SIDE
     ============================================================ -->

<div class="login-left">

    <div class="cap-dot dot1"></div>
    <div class="cap-dot dot2"></div>
    <div class="cap-dot dot3"></div>

    <div class="graduation-cap">
        🎓
    </div>

</div>


<!-- ============================================================
     RIGHT SIDE
     ============================================================ -->

<div class="login-right">


<h1 class="login-title">
AI Study Hub
</h1>


<p class="login-subtitle">
Learn • Practice • Test • Improve
</p>


<form method="POST">


<!-- USERNAME -->

<div class="login-input-box">

<span class="input-icon">
👤
</span>

<input
type="text"
name="username"
placeholder="Enter username"
required
>

</div>


<!-- PASSWORD -->

<div class="login-input-box">

<span class="input-icon">
🔒
</span>

<input
type="password"
name="password"
id="loginPassword"
placeholder="Enter password"
required
>

<span
class="password-eye"
id="passwordEye"
onclick="toggleLoginPassword()">

👁️

</span>

</div>


<!-- LOGIN BUTTON -->

<button
type="submit"
class="login-button">

🔐 LOGIN

</button>


</form>


<!-- CREATE ACCOUNT -->

<p class="create-account">

Don't have an account?

<a href="/signup">
Create Account
</a>

</p>


{% if error %}

<div class="error">

{{error}}

</div>

{% endif %}


</div>

</div>

</div>


<script>

/* ================================================================
   PASSWORD SHOW / HIDE
   ================================================================ */

function toggleLoginPassword() {

    const password =
        document.getElementById("loginPassword");

    const eye =
        document.getElementById("passwordEye");


    if (password.type === "password") {

        password.type = "text";

        eye.innerHTML = "🙈";

    }

    else {

        password.type = "password";

        eye.innerHTML = "👁️";

    }

}


/* ================================================================
   BACKGROUND GRADIENT CHANGE
   ================================================================ */

const colors = [

    "linear-gradient(135deg,#4f46e5,#06b6d4)",

    "linear-gradient(135deg,#7c3aed,#ec4899)",

    "linear-gradient(135deg,#2563eb,#9333ea)",

    "linear-gradient(135deg,#db2777,#7c3aed)",

    "linear-gradient(135deg,#0891b2,#6366f1)"

];


let c = 0;


setInterval(function() {

    c = (c + 1) % colors.length;

    document.body.style.background =
        colors[c];

}, 5000);

</script>

"""
# ================================================================
# 7. SIGNUP
# ================================================================

SIGNUP_HTML = COMMON_CSS + """

<div class="center">

<div class="card auth">

<div style="font-size:60px">✨</div>

<h1>Create Account</h1>

<form method="POST">

<input
name="username"
placeholder="Choose username"
required
>

<input
type="password"
name="password"
placeholder="Choose password"
required
>

<input
type="password"
name="confirm"
placeholder="Confirm password"
required
>

<button type="submit">
🚀 SIGN UP
</button>

</form>

<p>
Already have an account?
<a href="/" style="color:white;font-weight:bold;">
Login
</a>
</p>

{% if error %}
<div class="error">
{{error}}
</div>
{% endif %}

</div>
</div>

<script>

const colors=[
"linear-gradient(135deg,#06b6d4,#2563eb)",
"linear-gradient(135deg,#9333ea,#ec4899)",
"linear-gradient(135deg,#f59e0b,#ef4444)",
"linear-gradient(135deg,#059669,#06b6d4)"
];

let c=0;

setInterval(function(){
c=(c+1)%colors.length;
document.body.style.background=colors[c];
},5000);

</script>
"""

# ================================================================
# 8. DASHBOARD
# ================================================================

DASHBOARD_HTML = COMMON_CSS + """

<div class="container">

<div class="nav">

<div>
<h2>🎓 AI Study Hub</h2>
</div>

<div>
<span class="badge">
👤 {{username}}
</span>

<a class="btn" href="/logout">
Logout
</a>
</div>

</div>

<div class="card">

<h1>Welcome, {{username}} 👋</h1>

<p>
Choose a topic, study the material and test your knowledge.
</p>

<div class="grid">

<div class="card">
<div class="stat">{{topic_count}}</div>
<p style="text-align:center">Study Topics</p>
</div>

<div class="card">
<div class="stat">10</div>
<p style="text-align:center">MCQs / Topic</p>
</div>

<div class="card">
<div class="stat">100%</div>
<p style="text-align:center">Interactive Learning</p>
</div>

</div>

</div>

<br>

<h2>📚 Study Topics</h2>

<div class="grid">

{% for name,data in topics.items() %}

<div class="card topic-card">

<div class="icon">
{{data.icon}}
</div>

<h3>{{name}}</h3>

<p>
{{data.material[:170]}}...
</p>

<a class="btn" href="/study/{{name}}">
📖 Study
</a>

<a class="btn" href="/quiz/{{name}}">
📝 Quiz
</a>

</div>

{% endfor %}

</div>

<br>

<div class="card" style="text-align:center">

<h2>🤖 AI Score Prediction</h2>

<p>
Predict your possible performance based on study time and practice.
</p>

<a class="btn" href="/prediction">
Open Prediction
</a>

</div>

</div>

<script>

const colors=[
"linear-gradient(135deg,#312e81,#0891b2)",
"linear-gradient(135deg,#701a75,#be185d)",
"linear-gradient(135deg,#9a3412,#b91c1c)",
"linear-gradient(135deg,#065f46,#0369a1)",
"linear-gradient(135deg,#581c87,#1d4ed8)"
];

let c=0;

setInterval(function(){
c=(c+1)%colors.length;
document.body.style.background=colors[c];
},5000);

</script>
"""

# ================================================================
# 9. STUDY PAGE
# ================================================================

STUDY_HTML = COMMON_CSS + """

<div class="container">

<div class="nav">

<h2>{{data.icon}} {{topic}}</h2>

<a class="btn" href="/dashboard">
🏠 Dashboard
</a>

</div>

<div class="card">

<h1>📖 {{topic}}</h1>

<div class="material">
{{data.material}}
</div>

<br>

<div style="text-align:center">

<a class="btn" href="/quiz/{{topic}}">
📝 Start 10 Question Quiz
</a>

</div>

</div>

</div>
"""

# ================================================================
# QUIZ PAGE
# ================================================================

QUIZ_HTML = COMMON_CSS + """

<style>

/* ================================================================
   QUIZ PAGE
   ================================================================ */

.quiz-page {
    min-height: 100vh;
    padding: 30px;
    box-sizing: border-box;

    background:
        linear-gradient(
            135deg,
            #4f46e5,
            #7c3aed,
            #db2777
        );

    transition: background 1s ease;
}


/* Header */

.quiz-header {
    max-width: 1000px;
    margin: 0 auto 25px;

    display: flex;
    justify-content: space-between;
    align-items: center;

    padding: 20px 25px;

    border-radius: 20px;

    background: rgba(255,255,255,0.15);

    backdrop-filter: blur(15px);

    color: white;
}


.quiz-title {
    font-size: 30px;
    font-weight: 800;
    margin: 0;
}


.quiz-topic {
    margin-top: 5px;
    opacity: 0.9;
}


/* Timer */

.quiz-timer {
    background: white;
    color: #4f46e5;

    padding: 12px 20px;

    border-radius: 15px;

    font-size: 20px;
    font-weight: 800;

    min-width: 110px;

    text-align: center;
}


/* Questions */

.quiz-container {
    max-width: 1000px;
    margin: auto;
}


.question-card {

    background: rgba(255,255,255,0.96);

    border-radius: 22px;

    padding: 28px;

    margin-bottom: 22px;

    box-shadow:
        0 10px 30px rgba(0,0,0,0.12);
}


.question-number {

    display: inline-block;

    background: #6366f1;

    color: white;

    padding: 7px 13px;

    border-radius: 10px;

    font-size: 13px;

    font-weight: 700;

    margin-bottom: 12px;
}


.question-text {

    font-size: 20px;

    font-weight: 700;

    color: #1f2937;

    margin-bottom: 18px;

    line-height: 1.5;
}


/* ================================================================
   OPTIONS
   ================================================================ */

.option {

    width: 100%;

    box-sizing: border-box;

    padding: 16px 18px;

    margin: 10px 0;

    border-radius: 14px;

    border: 2px solid #e5e7eb;

    background: #f8fafc;

    color: #1f2937;

    cursor: pointer;

    font-size: 16px;

    transition: all 0.25s ease;
}


.option:hover {

    border-color: #6366f1;

    background: #eef2ff;

    transform: translateY(-2px);
}


/* Selected */

.option.selected {

    background:
        linear-gradient(
            135deg,
            #6366f1,
            #8b5cf6
        ) !important;

    color: white !important;

    border-color: #4f46e5 !important;

    box-shadow:
        0 8px 20px rgba(79,70,229,0.30);

    transform: scale(1.01);
}


/* Option letter */

.option-letter {

    display: inline-flex;

    align-items: center;
    justify-content: center;

    width: 30px;
    height: 30px;

    margin-right: 10px;

    border-radius: 50%;

    background: #e0e7ff;

    color: #4338ca;

    font-weight: 800;
}


.option.selected .option-letter {

    background: rgba(255,255,255,0.25);

    color: white;
}


/* Submit */

.submit-area {

    text-align: center;

    margin: 35px 0 50px;
}


.submit-button {

    border: none;

    padding: 16px 40px;

    border-radius: 16px;

    background: white;

    color: #4f46e5;

    font-size: 18px;

    font-weight: 800;

    cursor: pointer;

    box-shadow:
        0 10px 25px rgba(0,0,0,0.18);

    transition: 0.3s;
}


.submit-button:hover {

    transform: translateY(-4px) scale(1.03);

    box-shadow:
        0 15px 35px rgba(0,0,0,0.25);
}


/* ================================================================
   RESULT
   ================================================================ */

.result-box {

    max-width: 650px;

    margin: 30px auto;

    padding: 35px;

    text-align: center;

    border-radius: 25px;

    background: white;

    box-shadow:
        0 20px 50px rgba(0,0,0,0.25);

    display: none;
}


.result-title {

    font-size: 30px;

    font-weight: 800;

    color: #4f46e5;

    margin-bottom: 10px;
}


/* ================================================================
   SPEEDOMETER
   ================================================================ */

.speedometer {

    width: 300px;

    height: 155px;

    position: relative;

    margin: 25px auto 10px;

    overflow: hidden;

    border-radius: 300px 300px 0 0;

    background:

        conic-gradient(
            from 270deg,
            #ef4444 0deg,
            #f97316 45deg,
            #eab308 90deg,
            #84cc16 135deg,
            #22c55e 180deg,
            transparent 180deg
        );
}


/* Inner white area */

.speedometer::before {

    content: "";

    position: absolute;

    width: 225px;

    height: 112px;

    left: 37.5px;

    top: 43px;

    border-radius: 225px 225px 0 0;

    background: white;

    z-index: 2;
}


/* Needle */

.speed-needle {

    position: absolute;

    width: 5px;

    height: 115px;

    left: 147.5px;

    bottom: 0;

    background: #111827;

    border-radius: 10px;

    transform-origin: bottom center;

    transform: rotate(-90deg);

    transition: transform 1.5s ease;

    z-index: 5;
}


/* Center */

.speed-center {

    position: absolute;

    width: 22px;

    height: 22px;

    left: 139px;

    bottom: -2px;

    border-radius: 50%;

    background: #111827;

    z-index: 6;
}


/* Score */

.final-score {

    font-size: 38px;

    font-weight: 900;

    color: #4f46e5;

    margin-top: 8px;
}


/* Time */

.final-time {

    font-size: 19px;

    font-weight: 700;

    color: #374151;

    margin-top: 8px;
}


/* Message */

.result-message {

    font-size: 18px;

    font-weight: 600;

    margin-top: 15px;

    color: #4b5563;
}


/* Restart */

.restart-button {

    margin-top: 25px;

    padding: 13px 28px;

    border: none;

    border-radius: 13px;

    background: #4f46e5;

    color: white;

    font-weight: 700;

    font-size: 16px;

    cursor: pointer;
}


/* Mobile */

@media (max-width: 700px) {

    .quiz-page {
        padding: 15px;
    }

    .quiz-header {
        flex-direction: column;
        gap: 15px;
        text-align: center;
    }

    .quiz-title {
        font-size: 24px;
    }

    .question-card {
        padding: 20px;
    }

    .speedometer {
        transform: scale(0.85);
        margin-left: auto;
        margin-right: auto;
    }

}

</style>


<div class="quiz-page" id="quizPage">


<!-- ============================================================
     HEADER
     ============================================================ -->

<div class="quiz-header">

    <div>

        <h1 class="quiz-title">
            📝 Quiz
        </h1>

        <div class="quiz-topic">
            {{ topic }}
        </div>

    </div>


    <div class="quiz-timer" id="timer">
        ⏱️ 00:00
    </div>

</div>


<!-- ============================================================
     QUESTIONS
     ============================================================ -->

<div class="quiz-container" id="quizContainer">
</div>


<!-- ============================================================
     SUBMIT
     ============================================================ -->

<div class="submit-area" id="submitArea">

    <button
        class="submit-button"
        onclick="submitQuiz()">

        ✅ SUBMIT QUIZ

    </button>

</div>


<!-- ============================================================
     RESULT
     ============================================================ -->

<div class="result-box" id="resultBox">

    <div class="result-title">
        🎉 Quiz Completed!
    </div>


    <div class="speedometer">

        <div
            class="speed-needle"
            id="speedNeedle">
        </div>

        <div class="speed-center">
        </div>

    </div>


    <div
        class="final-score"
        id="finalScore">

        0%

    </div>


    <div
        class="final-time"
        id="finalTime">

        ⏱️ Time: 00:00

    </div>


    <div
        class="result-message"
        id="resultMessage">

    </div>


    <button
        class="restart-button"
        onclick="location.reload()">

        🔄 Try Again

    </button>

</div>


</div>


<script>

/* ================================================================
   QUESTIONS FROM FLASK
   ================================================================ */

const quizQuestions = {{ questions | tojson }};


/* ================================================================
   VARIABLES
   ================================================================ */

let answers = {};

let startTime = Date.now();

let timerSeconds = 0;

let timerRunning = true;


/* ================================================================
   RENDER QUESTIONS
   ================================================================ */

function loadQuiz() {

    const container =
        document.getElementById("quizContainer");

    container.innerHTML = "";


    quizQuestions.forEach(function(question, index) {

        const card =
            document.createElement("div");

        card.className = "question-card";


        const number =
            document.createElement("div");

        number.className = "question-number";

        number.innerText =
            "QUESTION " + (index + 1);


        const text =
            document.createElement("div");

        text.className = "question-text";

        text.innerText =
            question.question;


        card.appendChild(number);

        card.appendChild(text);


        /* Options */

        const options =
            question.options || [];


        options.forEach(function(option, optionIndex) {

            const optionDiv =
                document.createElement("div");

            optionDiv.className = "option";


            const letter =
                document.createElement("span");

            letter.className =
                "option-letter";


            const letters =
                ["A", "B", "C", "D"];


            letter.innerText =
                letters[optionIndex];


            const label =
                document.createElement("span");


            /*
             Supports both:
             "Artificial Intelligence"
             and
             {"value":"A","label":"Artificial Intelligence"}
            */

            if (typeof option === "object") {

                label.innerText =
                    option.label;

                optionDiv.dataset.value =
                    option.value;

            }

            else {

                label.innerText =
                    option;

                optionDiv.dataset.value =
                    optionIndex;

            }


            optionDiv.appendChild(letter);

            optionDiv.appendChild(label);


            /* Click */

            optionDiv.onclick = function() {

                /* Remove previous selection */

                const allOptions =
                    card.querySelectorAll(".option");


                allOptions.forEach(function(item) {

                    item.classList.remove("selected");

                });


                /* Highlight selected */

                optionDiv.classList.add("selected");


                /* Save answer */

                answers[index] =
                    optionDiv.dataset.value;

            };


            card.appendChild(optionDiv);

        });


        container.appendChild(card);

    });

}


/* ================================================================
   TIMER
   ================================================================ */

setInterval(function() {

    if (!timerRunning) {
        return;
    }


    timerSeconds++;


    const minutes =
        Math.floor(timerSeconds / 60);


    const seconds =
        timerSeconds % 60;


    const timeText =

        String(minutes).padStart(2, "0")
        + ":"
        +
        String(seconds).padStart(2, "0");


    document.getElementById("timer").innerText =
        "⏱️ " + timeText;

}, 1000);


/* ================================================================
   FORMAT TIME
   ================================================================ */

function formatTime(totalSeconds) {

    const minutes =
        Math.floor(totalSeconds / 60);


    const seconds =
        totalSeconds % 60;


    return (
        String(minutes).padStart(2, "0")
        +
        ":"
        +
        String(seconds).padStart(2, "0")
    );

}


/* ================================================================
   SUBMIT QUIZ
   ================================================================ */

async function submitQuiz() {

    if (!timerRunning) {
        return;
    }


    /* Stop timer */

    timerRunning = false;


    /* Disable submit */

    document.getElementById("submitArea").style.display =
        "none";


    /* Submit data */

    try {

        const response =
            await fetch("/submit_quiz", {

                method: "POST",

                headers: {
                    "Content-Type":
                        "application/json"
                },

                body: JSON.stringify({

                    topic: "{{ topic }}",

                    answers: answers

                })

            });


        if (!response.ok) {

            throw new Error(
                "Quiz submission failed"
            );

        }


        const result =
            await response.json();


        /*
         Expected backend response:

         {
             score: 8,
             total: 10
         }
        */

        const score =
            Number(result.score || 0);


        const total =
            Number(
                result.total ||
                quizQuestions.length
            );


        showResult(
            score,
            total,
            timerSeconds
        );


    }

    catch (error) {

        console.error(error);

        /*
         If backend returns an error,
         allow user to submit again.
        */

        timerRunning = true;

        document.getElementById("submitArea").style.display =
            "block";

        alert(
            "Unable to submit quiz. Please try again."
        );

    }

}


/* ================================================================
   SHOW FINAL RESULT
   ================================================================ */

function showResult(score, total, timeTaken) {

    const percentage =
        total > 0
        ? Math.round((score / total) * 100)
        : 0;


    /* Show result */

    document.getElementById("resultBox").style.display =
        "block";


    /* Score */

    document.getElementById("finalScore").innerText =
        score + " / " + total +
        "  (" + percentage + "%)";


    /* Time */

    document.getElementById("finalTime").innerText =
        "⏱️ Time Taken: " +
        formatTime(timeTaken);


    /* Message */

    let message = "";


    if (percentage >= 90) {

        message =
            "Excellent! 🌟 Great performance.";

    }

    else if (percentage >= 75) {

        message =
            "Very Good! 👍 Keep it up.";

    }

    else if (percentage >= 50) {

        message =
            "Good attempt! 📚 Keep practicing.";

    }

    else {

        message =
            "Keep learning and try again! 💪";

    }


    document.getElementById("resultMessage").innerText =
        message;


    /* ============================================================
       SPEEDOMETER NEEDLE
       ============================================================ */

    /*
       0%   = -90 degrees
       50%  = 0 degrees
       100% = 90 degrees
    */

    const angle =
        -90 + (percentage * 1.8);


    /*
       Small delay gives the needle
       a smooth animation.
    */

    setTimeout(function() {

        document.getElementById(
            "speedNeedle"
        ).style.transform =
            "rotate(" + angle + "deg)";

    }, 300);


    /* Scroll to result */

    setTimeout(function() {

        document.getElementById(
            "resultBox"
        ).scrollIntoView({
            behavior: "smooth",
            block: "center"
        });

    }, 400);

}


/* ================================================================
   START QUIZ
   ================================================================ */

loadQuiz();

</script>

"""

# ================================================================
# 11. PREDICTION PAGE
# ================================================================

PREDICTION_HTML = COMMON_CSS + """

<div class="container">

<div class="nav">

<h2>🤖 AI Score Prediction</h2>

<a class="btn" href="/dashboard">
🏠 Dashboard
</a>

</div>

<div class="card" style="max-width:650px;margin:auto">

<h1>📊 Predict Your Score</h1>

<p>
Enter your study information.
</p>

<form method="POST">

<label>
Study Days
</label>

<input
type="number"
name="days"
min="1"
max="30"
value="5"
required
>

<label>
Study Hours Per Day
</label>

<input
type="number"
name="hours"
min="1"
max="15"
value="3"
required
>

<label>
Practice Tests Completed
</label>

<input
type="number"
name="tests"
min="0"
max="30"
value="5"
required
>

<button type="submit">
🤖 Predict Score
</button>

</form>

{% if prediction is not none %}

<div class="card" style="margin-top:25px;text-align:center">

<h2>Predicted Score</h2>

<div style="font-size:60px;font-weight:bold">
{{prediction}}%
</div>

</div>

{% endif %}

</div>

</div>
"""

# ================================================================
# 12. LOGIN ROUTE
# ================================================================

@app.route("/", methods=["GET","POST"])
def login():

    if request.method == "POST":

        username = request.form.get("username","").strip()

        password = request.form.get("password","")

        conn = sqlite3.connect(DB_PATH)

        cur = conn.cursor()

        cur.execute(
            "SELECT id,password FROM users WHERE username=?",
            (username,)
        )

        user = cur.fetchone()

        conn.close()

        if user and check_password_hash(user[1],password):

            session["user_id"] = user[0]

            session["username"] = username

            return redirect("/dashboard")

        return render_template_string(
            LOGIN_HTML,
            error="Invalid username or password."
        )

    return render_template_string(
        LOGIN_HTML,
        error=None
    )


# ================================================================
# 13. SIGNUP ROUTE
# ================================================================

@app.route("/signup", methods=["GET","POST"])
def signup():

    if request.method == "POST":

        username = request.form.get("username","").strip()

        password = request.form.get("password","")

        confirm = request.form.get("confirm","")

        if len(username) < 3:

            return render_template_string(
                SIGNUP_HTML,
                error="Username must contain at least 3 characters."
            )

        if len(password) < 4:

            return render_template_string(
                SIGNUP_HTML,
                error="Password must contain at least 4 characters."
            )

        if password != confirm:

            return render_template_string(
                SIGNUP_HTML,
                error="Passwords do not match."
            )

        try:

            conn = sqlite3.connect(DB_PATH)

            cur = conn.cursor()

            cur.execute(
                "INSERT INTO users(username,password) VALUES(?,?)",
                (
                    username,
                    generate_password_hash(password)
                )
            )

            conn.commit()

            conn.close()

            return redirect("/")

        except sqlite3.IntegrityError:

            return render_template_string(
                SIGNUP_HTML,
                error="Username already exists."
            )

    return render_template_string(
        SIGNUP_HTML,
        error=None
    )


# ================================================================
# 14. DASHBOARD ROUTE
# ================================================================

@app.route("/dashboard")
def dashboard():

    if "user_id" not in session:

        return redirect("/")

    return render_template_string(
        DASHBOARD_HTML,
        username=session.get("username"),
        topics=topics,
        topic_count=len(topics)
    )


# ================================================================
# 15. STUDY ROUTE
# ================================================================

@app.route("/study/<path:topic>")
def study(topic):

    if "user_id" not in session:

        return redirect("/")

    if topic not in topics:

        return "Topic not found",404

    return render_template_string(
        STUDY_HTML,
        topic=topic,
        data=topics[topic]
    )


# ================================================================
# 16. QUIZ ROUTE
# ================================================================

@app.route("/quiz/<path:topic>")
def quiz(topic):

    if "user_id" not in session:

        return redirect("/")

    if topic not in topics:

        return "Topic not found",404

    # Convert questions into browser-safe dictionaries.
    questions_for_browser = []

    for question,options,correct in topics[topic]["questions"]:

        questions_for_browser.append({

            "question":question,

            "options":options

        })

    return render_template_string(

        QUIZ_HTML,

        topic=topic,

        questions=questions_for_browser

    )


# ================================================================
# 17. QUIZ SUBMISSION
# ================================================================

@app.route("/submit_quiz",methods=["POST"])
def submit_quiz():

    try:

        if "user_id" not in session:

            return jsonify({
                "success":False,
                "error":"Your session expired. Please login again."
            }),401

        data = request.get_json(silent=True)

        if not isinstance(data,dict):

            return jsonify({
                "success":False,
                "error":"Invalid quiz request."
            }),400

        topic = data.get("topic")

        answers = data.get("answers",{})

        if topic not in topics:

            return jsonify({
                "success":False,
                "error":"Invalid topic."
            }),400

        if not isinstance(answers,dict):

            return jsonify({
                "success":False,
                "error":"Invalid answers."
            }),400

        questions = topics[topic]["questions"]

        score = 0

        for i,q in enumerate(questions):

            correct_answer = int(q[2])

            user_answer = answers.get(str(i))

            if user_answer is None:
                continue

            try:

                if int(user_answer) == correct_answer:
                    score += 1

            except:

                continue

        total = len(questions)

        percentage = round(
            (score/total)*100,
            2
        ) if total else 0

        conn = sqlite3.connect(DB_PATH)

        cur = conn.cursor()

        cur.execute("""
        INSERT INTO results(
            user_id,
            topic,
            score,
            total,
            percentage
        )
        VALUES(?,?,?,?,?)
        """,(
            session["user_id"],
            topic,
            score,
            total,
            percentage
        ))

        conn.commit()

        conn.close()

        return jsonify({

            "success":True,

            "score":score,

            "total":total,

            "percentage":percentage

        })

    except Exception as e:

        print(
            "QUIZ SUBMISSION ERROR:",
            repr(e)
        )

        return jsonify({

            "success":False,

            "error":
            "Quiz server error: "
            + str(e)

        }),500


# ================================================================
# 18. PREDICTION
# ================================================================

@app.route("/prediction",methods=["GET","POST"])
def prediction():

    if "user_id" not in session:

        return redirect("/")

    predicted = None

    if request.method == "POST":

        days = float(
            request.form.get("days",5)
        )

        hours = float(
            request.form.get("hours",3)
        )

        tests = float(
            request.form.get("tests",5)
        )

        # Keep inputs in model range.
        days = max(1,min(days,10))
        hours = max(1,min(hours,65))
        tests = max(0,min(tests,90))

        features = pd.DataFrame(
            [[days,hours,tests]],
            columns=["days","hours","tests"]
        )

        predicted = round(
            float(model.predict(features)[0]),
            2
        )

        predicted = max(
            0,
            min(predicted,100)
        )

    return render_template_string(

        PREDICTION_HTML,

        prediction=predicted

    )


# ================================================================
# 19. LOGOUT
# ================================================================

@app.route("/logout")
def logout():

    session.clear()

    return redirect("/")


# ================================================================
# 20. ERROR HANDLERS
# ================================================================

@app.errorhandler(500)
def internal_error(error):

    print(
        "FLASK INTERNAL ERROR:",
        repr(error)
    )

    if request.path == "/submit_quiz":

        return jsonify({

            "success":False,

            "error":
            "Server error while submitting quiz. "
            "Please try again."

        }),500

    return """
    <h1>Application Error</h1>
    <p>Please return to the dashboard and try again.</p>
    """,500


@app.errorhandler(404)
def not_found(error):

    return """
    <h1>Page Not Found</h1>
    <p><a href="/dashboard">Go to Dashboard</a></p>
    """,404


# ================================================================
# 21. FIND FREE PORT
# ================================================================

def find_free_port():

    sock = socket.socket(
        socket.AF_INET,
        socket.SOCK_STREAM
    )

    sock.bind(("127.0.0.1",0))

    port = sock.getsockname()[1]

    sock.close()

    return port


PORT = find_free_port()

print("="*65)

print("🎓 AI STUDY HUB")

print("="*65)

print(
    "📚 Topics:",
    len(topics)
)

print(
    "📝 Questions per topic: 10"
)

print(
    "💾 Database:",
    DB_PATH
)

print(
    "🌐 Port:",
    PORT
)

print("="*65)


# ================================================================
# 22. START FLASK
# ================================================================

def run_app():

    app.run(
        host="0.0.0.0",
        port=PORT,
        debug=False,
        use_reloader=False,
        threaded=True
    )


thread = threading.Thread(
    target=run_app,
    daemon=True
)

thread.start()

time.sleep(3)


# ================================================================
# 23. CREATE COLAB WEBSITE LINK
# ================================================================

from IPython.display import display, HTML
from google.colab import output

try:
    url = output.eval_js(
        f"google.colab.kernel.proxyPort({PORT})"
    )

    print("=" * 65)
    print("🎉 AI STUDY HUB IS LIVE!")
    print("=" * 65)
    print("\n🌐 Website URL:")
    print(url)
    print("\n⚠️ Keep this Colab runtime running while using the website.")

    display(HTML(f"""
    <div style="
        padding:30px;
        border-radius:20px;
        background:linear-gradient(135deg,#4f46e5,#06b6d4,#9333ea);
        text-align:center;
        color:white;
        font-family:Arial;
        margin:20px 0;
    ">

        <h1>🎓 AI STUDY HUB</h1>

        <p style="font-size:16px;">
            Login → Dashboard → Study Material → Quiz → Score
        </p>

        <a href="{url}"
           target="_blank"
           style="
             display:inline-block;
             padding:16px 30px;
             background:white;
             color:#4f46e5;
             border-radius:12px;
             text-decoration:none;
             font-size:18px;
             font-weight:bold;
           ">
           🚀 OPEN AI STUDY HUB
        </a>

        <p style="margin-top:20px;font-size:13px;">
            Keep Google Colab running while the website is being used.
        </p>

    </div>
    """))

except Exception as e:
    print("❌ Website link generation error:", e)
    print("Flask is running on port:", PORT)

# ================================================================
# END
# ================================================================
```
